#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
ncaa_tidy |> 
    select(where(is.numeric)) |> # select for only numerical variables
    pivot_longer(
        cols = everything(),
        names_to = "variable",
        values_to = "value") |> # reshape data into variable and value pairs 
    group_by(variable) |> # group observations by their original variable
    summarize(
        mean = mean(value, na.rm = TRUE),
        min = min(value, na.rm = TRUE),
        max = max(value, na.rm = TRUE)) # calculate mean, minimum, and maximum for each variable
```

```r
options(repr.plot.width = 12, repr.plot.height = 6)
# visualize the distribution of tournament wins
tourney_wins_plot <- ncaa_tidy |>
    ggplot(aes(x = tourney_wins)) + # map tournament wins to x axis
    geom_bar(fill = "orange") + # display the frequency of each victory count
    labs(
        x = "Number of Tournament Victories",
        y = "Number of Team Appearances",
        title = "Distribution of NCAA Tournament Victories") + # add descriptive titles and axis
    scale_x_continuous(breaks = 0:6) + # display each possible x value
    theme(
        plot.title = element_text(size = 18, hjust = 0.5), # center and enlarge title
        axis.title = element_text(size = 14), # enlarge axes titles
        axis.text = element_text(size = 12) # enlarge axis labels
        ) 
tourney_wins_plot
```

```r
options(repr.plot.width = 12, repr.plot.height = 6)
# visualize the proportion of tournament seedings
seed_proportion_plot <- ncaa_tidy |>
    filter(seed != 0) |> # remove 0 seeding from 1983
    count(seed) |> # count the number of teams per seed
    mutate(proportion = n / sum(n)) |> # calculate each seeds proportion
    ggplot(aes(x = factor(seed), y = proportion)) + # treat seeds as categories and map proportion to y axis
    geom_bar(stat = "identity", fill = "steelblue") + # plot the proportions as bars 
    labs(
        x = "Tournament Seed",
        y = "Proportion of Tournament Appearances",
        title = "Relative Proportion of Tournament Seeds") + # add labels and titles
    scale_y_continuous(labels = scales::percent) + # display proportion as a percentage
    theme(
        plot.title = element_text(size = 18, hjust = 0.5), # center and enlarge title
        axis.title = element_text(size = 14), # enlarge axes titles
        axis.text = element_text(size = 12) # enlarge axes labels
        )
seed_proportion_plot
```

The first visualization illustrates a positive association between regular season win percentage and tournament success. There is an approximately 20-percentage-point increase in median regular season win percentage from teams with 0 victories to 6 victories. The overlapping distributions suggest teams with similar regular season records can experience different tournament outcomes. Thus suggests that a regular season win percentage may be a useful predictor in a K-NN regression model, however, it cannot fully explain variation in tournament performance. 

```r
# visualize relationship between regular season performance and tournament wins
reg_wins_plot <- ncaa_tidy |>
    mutate(reg_wins_performance = reg_wins_pct / 100) |> # convert to true percentage value
    ggplot(aes(x = factor(tourney_wins), y = reg_wins_performance)) + # map predictor and response variables
    geom_boxplot(fill = "orange") + # plot observations to view median, spread, outliers
    labs(
        x = "Number of Tournament Victories",
        y = "Regular Season Win Percentage",
        title = "Regular Season Win Percentage by Tournament Victories") + # add title and axes labels
    scale_y_continuous(labels = scales::percent) + # display y axis as percentage
    theme(
        plot.title = element_text(size = 18, hjust = 0.5), # center and enlarge title
        axis.title = element_text(size = 14), # enlarge axes titles
        axis.text = element_text(size = 12) # enlarge axes labels
        )
reg_wins_plot
```

The second visualization demonstrates a strong negative association between descending tournament seed and average tournament victories. First-seeded teams average about 3.5 victories, compared to fewer than one victory in teams seeded sixth or lower. This relationship suggests that tournament seeding may be particularly informative as a predictor in a K-NN regression model considering how well it appears to reflect differences in tournament performance. 

```r
# visualize relationship between tournament seed and mean number of tournament victories
seed_wins_plot <- ncaa_tidy |>
    filter(seed != 0) |> # exclude 1983 opening round 0 seeding
    group_by(seed) |> 
    summarize(
        mean_wins = mean(tourney_wins, na.rm = TRUE)
        ) |> # group by seed and calculate average tournament win
    ggplot(aes(x = seed, y = mean_wins)) + # treat seed as category
    geom_bar(stat = "identity", fill = "steelblue") + # plot observations to view median, spread, outliers
    scale_x_continuous(breaks = 1:16) + # include every seed as label
    labs(
        x = "Tournament Seed",
        y = "Mean Number of Tournament Victories",
        title = "Average Tournament Victories by Tournament Seed") + # add titles and labels
    theme(
        plot.title = element_text(size = 18, hjust = 0.5), # center and enlarge title
        axis.title = element_text(size = 14), # enlarge axes titles
        axis.text = element_text(size = 12) # enlarge axes labels
        )
seed_wins_plot
```

The third and final visualization suggests a positive association between conference placement and tournment success. First-place teams average just over 1 tournament victory compared to about 0.5 with ninth-place teams. While the relationship is not perfectly consistent, the overall trend suggests that teams that perform better within their conferences, typically do better in the tournament.  Therefore, conference ranking will likely provide useful predictive information in a K-NN regression model, but differences in conference size and competitive strength will limit direct comparisons between rankings. 

```r
# visualize relationship between conference rank and tournament victories
conf_rank_plot <- ncaa_tidy |>
    group_by(conf_rank) |>
    summarize(
        mean_wins = mean(tourney_wins, na.rm = TRUE)
        ) |> # calculate mean tourney victories for each ranking
    ggplot(aes(x = conf_rank, y = mean_wins)) + # map conference rank and mean tourney wins
    geom_line(color = "mediumpurple", linewidth = 1) +
    geom_point(color = "mediumpurple", size = 3) + # plot observations
    scale_x_reverse(breaks = 10:1) + # include all seeds as labels in descending order
    labs(
        x = "Conference Ranking",
        y = "Mean Number of Tournament Victories",
        title = "Average Tournament Victories by Conference Ranking"
    ) + # add title and labels
    theme(
        plot.title = element_text(size = 18, hjust = 0.5), # center and enlarge title
        axis.title = element_text(size = 14), # enlarge axes titles
        axis.text = element_text(size = 12) # enlarge axes labels
        )
conf_rank_plot
```

This cross validation will serve in place of a seperate validation set. After tuning, the final model will be fitted using the training set and evaluated on unseen testing data. 

Finally, model performance will be assessed using root mean squared prediction error (RMSPE).
